```python
## assign values to variables
murders_chicago, pop_chicago = 762, 2_705_000
murders_nyc, pop_nyc = 334, 8_538_000
murders_sj, pop_sj = 47, 1_025_000
```

```python
## calculate rate for nyc
(murders_nyc/pop_nyc) * 100_000
```

```python
## place result into variable
murder_rate_nyc = (murders_nyc/pop_nyc) * 100_000
```

```python
print(murder_rate_nyc)
```

```python
## calculate rate for chicago
(murders_chicago/pop_chicago) * 100_000
```

```python
## place result into variable
murder_rate_chicago = (murders_chicago/pop_chicago) * 100_000
```

```python
print(murder_rate_chicago)
```

```python
## calculate rate for san jose
(murders_sj/pop_sj) * 100_000
```

```python
## place result into variable
murder_rate_sj = (murders_sj/pop_sj) * 100_000
```

```python
print(murder_rate_sj)
```

```python
## write sentence using f-strings that notes which city has the lowest murder rate
print(f"New York City has a murder rate of {murder_rate_nyc:.2F} per 100K while other cities such as San Jose and Chicago have higher murder rates of {murder_rate_sj:.2F} and {murder_rate_chicago:.2F} per 100K, respectively.")
```
